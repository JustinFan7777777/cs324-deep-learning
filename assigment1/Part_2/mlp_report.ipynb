## Part II: batch gradient descent

The default setting uses one full training-set batch per epoch.

```python
batch_model, batch_history = train(
    dnn_hidden_units="20",
    learning_rate=0.01,
    max_steps=1500,
    eval_freq=10,
    batch_size=None,
    random_seed=42,
)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(batch_history["steps"], batch_history["train_accuracy"], label="Training accuracy")
plt.plot(batch_history["steps"], batch_history["test_accuracy"], label="Test accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training and test accuracy: full-batch gradient descent")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(batch_history["steps"], batch_history["train_loss"], label="Training loss")
plt.plot(batch_history["steps"], batch_history["test_loss"], label="Test loss")
plt.xlabel("Epoch")
plt.ylabel("Cross-entropy loss")
plt.title("Training and test loss: full-batch gradient descent")
plt.legend()
plt.grid(True)
plt.show()
```

## Part III: stochastic and mini-batch gradient descent

`batch_size=1` is SGD. The other values show how mini-batch size changes the learning curves. A value of `None` means the complete training set.

```python
batch_sizes = [1, 8, 16, 32, 64, 128, None]
batch_results = {}

for batch_size in batch_sizes:
    label = "full batch" if batch_size is None else f"batch size {batch_size}"
    _, batch_results[label] = train(
        dnn_hidden_units="20",
        learning_rate=0.01,
        max_steps=1500,
        eval_freq=10,
        batch_size=batch_size,
        random_seed=42,
    )
```

```python
plt.figure(figsize=(10, 6))
for label, history in batch_results.items():
    plt.plot(history["steps"], history["test_accuracy"], label=label)
plt.xlabel("Epoch")
plt.ylabel("Test accuracy")
plt.title("Influence of batch size on test accuracy")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
for label, history in batch_results.items():
    plt.plot(history["steps"], history["test_loss"], label=label)
plt.xlabel("Epoch")
plt.ylabel("Test cross-entropy loss")
plt.title("Influence of batch size on test loss")
plt.legend()
plt.grid(True)
plt.show()
```

```python
summary = {
    label: history["test_accuracy"][-1]
    for label, history in batch_results.items()
}
for label, final_accuracy in summary.items():
    print(f"{label}: {final_accuracy:.2%}")
```
